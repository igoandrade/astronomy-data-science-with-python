## 1) Import the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage.io import imread # skimage is one of the lib for image processing (cv2 is another example)
```

## 2) Read the image

```python
# Create variable to store the filepath
file_path = '/content/istockphoto-470985038-612x612.jpg'

# Read the file path and save the numpy array that it returns
img_array = imread(file_path)
```

```python
# Check the type of img_array
type(img_array)
```

## 3) Understand the properties of this image array

```python
# Check the shape of the image array
print(img_array.shape)

# Check the data type of the image array
print(img_array.dtype)

# Print the minimum and maximum value in this array
print(img_array.min())
print(img_array.max())
```

## 4) Display the image using matplotlib

```python
# display image from img_array
plt.figure(figsize=(8,10))       # Adjust the figure size
plt.imshow(img_array)            # Display the array as image
plt.title('Night Sky!',          # Set title
          color='royalblue',
          weight='bold',
          fontsize=15)
plt.axis('off')                  # Turn off the axis
plt.show()                       # To avoid the output from previous line
```

## 5) Convert the image into grayscale and display it

```python
# Read the image by setting as_gray = True
gray_img_array = imread(file_path, as_gray=True)
plt.imshow(gray_img_array, cmap='gray')
plt.title(f'The shape of 2D image:- {gray_img_array.shape}')
plt.show()
```

## 6) Display the grayscale and original image side-by-side using subplot

```python
# Adjust the figuresize based on the subplot
plt.figure(figsize=(8, 10))

# Image axes on the left (position 1)
plt.subplot(1, 2, 1)
plt.imshow(img_array)
plt.title('Original', color='maroon')
plt.axis('off')


# Image axes on the right (position 2)
plt.subplot(1, 2, 2)
plt.imshow(gray_img_array, cmap='gray')
plt.title('Grayscale', color='k')
plt.axis('off')

# Adjusting the axes in one figure automatically!
plt.tight_layout()

# To save the figure as png file
plt.savefig('first_subplot.png')
```

## 7) Cropping the arrays and visualize it

```python
# Check the current height and width of the image which will allow you to crop later
plt.imshow(img_array)
plt.show()
```

```python
# Crop the Orion Constellation
plt.imshow(img_array[40:247, 0:200, :]) # slicing array[row, col, depth] or array[height, width, channels]
plt.axis('off')
plt.title('Orion Constellation', color='maroon', weight='bold')
plt.show()
```
