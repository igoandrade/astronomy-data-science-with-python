```python
# Load the libraries
from skimage.io import imread
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Get the image path
image_path = '/content/black-hole-4118711_1920.jpg'
```

```python
# Get the image array using imread
image_array = imread(image_path)
```

```python
# Check the shape of the array
image_array.shape
```

```python
# Check the dtype
image_array.dtype
```

```python
# Extract the channels
r_channel = image_array[:, :, 0]
g_channel = image_array[:, :, 1]
b_channel = image_array[:, :, 2]
```

```python
# Check the shapes of each of these channels
print(r_channel.shape)
print(g_channel.shape)
print(b_channel.shape)
```

```python
# Supbplot of three axes in one line (horizontal) (1, 4, position)

titles = ('Original', 'Red Channel', 'Green Channel', 'Blue Channel')
titles_color = ('k', 'r', 'g', 'b')
arrays_plot = (image_array, r_channel, g_channel, b_channel)

plt.figure(figsize=(10, 5))
for index, array_ in enumerate(arrays_plot):
  plt.subplot(1, 4, index+1)
  plt.imshow(array_, cmap='gray')
  plt.title(titles[index], color = titles_color[index], weight='bold')
  plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# Merge the image in BGR format and GBR format and display as side by side image
bgr_image = np.stack([b_channel, g_channel, r_channel], axis=-1)
gbr_image = np.stack([g_channel, b_channel, r_channel], axis=-1)

titles1 = ('RGB Image', 'BGR Image', 'GBR Image')
titles1_color = ('black', 'maroon', 'royalblue')

plt.figure(figsize=(10, 8))
for index, image in enumerate((image_array, bgr_image, gbr_image)):
  plt.subplot(1, 3, index+1)
  plt.imshow(image)
  plt.title(titles1[index], color=titles_color[index], weight='bold')
  plt.axis('off')

plt.tight_layout()
plt.show()
```
