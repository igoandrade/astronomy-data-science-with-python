# Imagens com Python

```python
# Bibliotecas importadas

from skimage.io import imread
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Caminho para imagem

image_path = '../content/istockphoto-470985038-612x612.jpg'
```

```python
img_array = imread(image_path)
```

```python
print(f"(rows, columns, depth) = {img_array.shape}")
```

```python
# Display image using matplotlib


plt.figure(figsize=(10, 8))
plt.imshow(img_array)

plt.title("Night Sky, Pleiades and Orion")

plt.show()
```

```python
# COnverter imagem para escala de cinza

gray_img_array = imread(image_path, as_gray=True)

plt.figure(figsize=(10, 8))
plt.imshow(gray_img_array, cmap='gray')
plt.title(f'Night Sky in grayscale (2D): {gray_img_array.shape}')
plt.show()
```

```python

plt.figure(figsize=(10, 8))
plt.subplot(1, 2, 1)
plt.imshow(img_array)
plt.title("Original image")

plt.subplot(1, 2, 2)
plt.imshow(gray_img_array, cmap='gray')
plt.title("Grayscale image")

plt.tight_layout()
plt.show()
```

```python
# Crop the Orion Constalation

plt.imshow(img_array[40:240, :200, :])
plt.title('Orion Constalation')
plt.axis('off')

plt.show()
```

```python
image2_path = '../content/black-hole-4118711_1920.jpg'
```

```python
img2_array = imread(image2_path)
img2_array.shape
```

```python
# Extrair canais RGB

r_channel = img2_array[:, :, 0]
g_channel = img2_array[:, :, 1]
b_channel = img2_array[:, :, 2]
```

```python
# Visualização por canal RGB

titles = (
    'Original', 'Red Channel', 'Green Channel', 'Blue Channel'
)
cmaps = ('', 'Reds', 'Greens', 'Blues')
arrays_plot = (img2_array, r_channel, g_channel, b_channel)


plt.figure(figsize=(10, 9))
for index, array_plot in enumerate(arrays_plot):
    plt.subplot(1, 4, index+1)
    plt.title(titles[index])
    if index == 0:
        plt.imshow(array_plot)
    else:
        plt.imshow(array_plot, cmap=cmaps[index])
    plt.axis('off')

plt.tight_layout()
plt.show()
```
